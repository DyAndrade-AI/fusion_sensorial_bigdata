### 🖥️ Interfaz Gráfica de Usuario (GUI) Interactiva para la Presentación

Ejecuta la siguiente celda para inicializar un panel de control visual desarrollado con `ipywidgets`. Esta interfaz te permitirá simular de manera dinámica los atributos de música y color frente a los sinodales, mostrando las recomendaciones de inmediato mediante un diseño estructurado en HTML.

````python
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
import subprocess
import json
import os

# Asegurar que estamos en el directorio correcto
if os.path.exists('/content/fusion_sensorial'):
    os.chdir('/content/fusion_sensorial')

# --- ESTILOS DE LA INTERFAZ ---
titulo_html = """
<div style="background-color:#1e293b; padding:15px; border-radius:10px; margin-bottom:15px; text-align:center;">
    <h2 style="color:#f8fafc; margin:0; font-family:sans-serif;">🔮 Fusión Sensorial: Consola Interactiva</h2>
    <p style="color:#94a3b8; margin:5px 0 0 0; font-family:sans-serif;">Ajusta los parámetros sensoriales en vivo para buscar tu destino ideal</p>
</div>
"""

# --- ELEMENTOS DE CONTROL (WIDGETS) ---
style = {'description_width': 'initial'}

bpm_slider = widgets.IntSlider(value=120, min=60, max=180, step=5, description='Tempo (BPM):', style=style, layout=widgets.Layout(width='90%'))
energy_slider = widgets.FloatSlider(value=0.7, min=0.0, max=1.0, step=0.05, description='Energía:', style=style, layout=widgets.Layout(width='90%'))
dance_slider = widgets.FloatSlider(value=0.6, min=0.0, max=1.0, step=0.05, description='Bailabilidad:', style=style, layout=widgets.Layout(width='90%'))
valence_slider = widgets.FloatSlider(value=0.5, min=0.0, max=1.0, step=0.05, description='Valencia (Alegría):', style=style, layout=widgets.Layout(width='90%'))

color_picker = widgets.ColorPicker(concise=False, description='Color Hex:', value='#FF5733', style=style)
presupuesto_slider = widgets.FloatSlider(value=0.0, min=0.0, max=1.0, step=0.1, description='Importancia Presupuesto:', style=style, layout=widgets.Layout(width='90%'))
joyas_checkbox = widgets.Checkbox(value=False, description='Filtrar solo Joyas Ocultas', style=style)

boton_recomendar = widgets.Button(
    description='🚀 Obtener Recomendación',
    disabled=False,
    button_style='success',
    tooltip='Presiona para calcular afinidades con destinos',
    icon='search',
    layout=widgets.Layout(width='95%', height='45px', margin='15px 0px')
)

salida_resultados = widgets.Output()

# --- DISPOSICIÓN VISUAL (LAYOUT) ---
columna_izquierda = widgets.VBox([
    widgets.HTML("<h4>🎵 Parámetros Musicales</h4>"),
    bpm_slider, energy_slider, dance_slider, valence_slider
], layout=widgets.Layout(padding='10px', border='1px solid #cbd5e1', border_radius='8px', margin='5px', width='48%'))

columna_derecha = widgets.VBox([
    widgets.HTML("<h4>🎨 Parámetros Cromáticos y Filtros</h4>"),
    color_picker,
    widgets.HTML("<br>"),
    presupuesto_slider,
    joyas_checkbox
], layout=widgets.Layout(padding='10px', border='1px solid #cbd5e1', border_radius='8px', margin='5px', width='48%'))

panel_controles = widgets.HBox([columna_izquierda, columna_derecha])

# --- FUNCIÓN DE RESPUESTA AL CLICK ---
def procesar_recomendacion(b):
    with salida_resultados:
        clear_output()
        display(HTML("<div style='text-align:center; padding:20px;'><h4 style='color:#3b82f6;'>⚙️ Ejecutando algoritmo de recomendación...</h4></div>"))

        # Construir comando
        cmd = [
            "python", "-m", "src.main",
            "--bpm", str(bpm_slider.value),
            "--energy", str(energy_slider.value),
            "--danceability", str(dance_slider.value),
            "--valence", str(valence_slider.value),
            "--color", color_picker.value,
            "--presupuesto", str(presupuesto_slider.value),
            "--top", "3"
        ]
        if joyas_checkbox.value:
            cmd.append("--solo-joyas")

        try:
            # Ejecutar script y capturar salida
            resultado = subprocess.run(cmd, capture_output=True, text=True, check=True)

            # Formatear la salida estándar para mostrarla limpia en la interfaz
            salida_limpia = resultado.stdout.replace("\n", "<br>")
            display(HTML(f"""
            <div style="background-color:#f1f5f9; border-left: 5px solid #10b981; padding:15px; border-radius:4px; font-family:monospace; margin-top:15px; color:#1e293b;">
                {salida_limpia}
            </div>
            """))
        except subprocess.CalledProcessError as err:
            # Captura de errores de ejecución
            error_msg = err.stderr if err.stderr else "No se pudo ejecutar src.main."
            # Si ocurre un ModuleNotFoundError, ofrecer el comando directo con el PATH ajustado
            if "ModuleNotFoundError" in error_msg:
                try:
                    cmd_alternativo = ["python3", "src/main.py"] + cmd[3:]
                    resultado_alt = subprocess.run(cmd_alternativo, capture_output=True, text=True, check=True)
                    salida_limpia = resultado_alt.stdout.replace("\n", "<br>")
                    display(HTML(f"""
                    <div style="background-color:#f1f5f9; border-left: 5px solid #10b981; padding:15px; border-radius:4px; font-family:monospace; margin-top:15px; color:#1e293b;">
                        {salida_limpia}
                    </div>
                    """))
                    return
                except Exception as ex:
                    error_msg += f"\n[Intento fallback fallido]: {str(ex)}"

            display(HTML(f"""
            <div style="background-color:#fef2f2; border-left: 5px solid #ef4444; padding:15px; border-radius:4px; font-family:monospace; margin-top:15px; color:#991b1b;">
                <strong>❌ Error al ejecutar el script del recomendador:</strong><br><br>{error_msg.replace('\n', '<br>')}
            </div>
            """))

boton_recomendar.on_click(procesar_recomendacion)

# --- RENDERIZADO FINAL DE LA GUI ---
display(HTML(titulo_html))
display(panel_controles)
display(boton_recomendar)
display(salida_resultados)
````

### 🎤 Intérprete Narrativo Sensorial en Vivo para la Exposición
Usa esta celda interactiva durante la presentación. Además de mostrar las recomendaciones físicas, generará un párrafo narrativo que traduce matemáticamente tus selecciones sensoriales al lenguaje humano en tiempo real.

````python
# @title 🎛️ Consola Narrativa Interactiva
BPM = 140 # @param {type:"slider", min:60, max:180, step:5}
Energia = 0.85 # @param {type:"slider", min:0.0, max:1.0, step:0.05}
Bailabilidad = 0.8 # @param {type:"slider", min:0.0, max:1.0, step:0.05}
Valencia_Alegria = 0.75 # @param {type:"slider", min:0.0, max:1.0, step:0.05}
Color_Hex = "#1F6966" # @param {type:"string"}
Presupuesto_Importancia = 0.5 # @param {type:"slider", min:0.0, max:1.0, step:0.05}
Solo_Joyas_Ocultas = False # @param {type:"boolean"}

# --- TRADUCTOR NARRATIVO --- (Lógica espejo del módulo sensory.py)
import matplotlib.colors as mcolors
import math

# 1. Análisis de música
bpm_n = clip_val = max(0.0, min(1.0, (BPM - 60) / 120))
act_musica = 0.35 * bpm_n + 0.35 * Energia + 0.30 * Bailabilidad
pos_musica = Valencia_Alegria

# 2. Análisis de color
try:
    rgb = mcolors.to_rgb(Color_Hex)
    hsv = mcolors.rgb_to_hsv(rgb)
    h, s, v = hsv[0] * 360, hsv[1], hsv[2]
    # Calidez basada en la cercanía al naranja (30 grados)
    calidez_c = (1.0 + math.cos(math.radians(h - 30.0))) / 2.0
    if s < 0.1:  # Atenuación de calidez para colores neutros/grises
        calidez_c = 0.5
    act_color = 0.5 * s + 0.3 * calidez_c + 0.2 * v
    pos_color = 0.5 * v + 0.5 * calidez_c
except Exception as e:
    act_color, pos_color, calidez_c = 0.5, 0.5, 0.5

# Fusion Tardía
activacion_final = 0.5 * act_musica + 0.5 * act_color
positividad_final = 0.5 * pos_musica + 0.5 * pos_color

# Generar Perfil Narrativo
intro = f"### 🔮 ANÁLISIS DE TU PERFIL SENSORIAL\n"
intro += f"El usuario ha introducido un ritmo de **{BPM} BPM** con un color **{Color_Hex}**.\n\n"

# Interpretación de la energía
if activacion_final > 0.7:
    energia_txt = "una vibra de alta intensidad, ritmo acelerado y un fuerte deseo de movimiento físico"
elif activacion_final > 0.4:
    energia_txt = "un estado de ánimo equilibrado, dinámico pero moderado, ideal para actividades activas sin llegar al cansancio extreme"
else:
    energia_txt = "una actitud contemplativa, pacífica, de profunda relajación y bajo consumo de energía"

# Interpretación del color y calidez
if calidez_c > 0.7:
    calidez_txt = "tonalidades cálidas y acogedoras que incitan a la cercanía social y al confort"
elif calidez_c > 0.4:
    calidez_txt = "tonalidades templadas y balanceadas"
else:
    calidez_txt = "tonalidades frías y refrescantes, asociadas con la frescura, el agua y la introspección personal"

# Interpretación de la positividad
if positividad_final > 0.65:
    animo_txt = "optimismo radiante y felicidad"
elif positividad_final > 0.45:
    animo_txt = "una estabilidad emocional tranquila"
else:
    animo_txt = "nostalgia reflexiva o melancolía pacífica"

final_text = f"{intro}*\"Buscas un lugar que transmita **{animo_txt}**, con **{energia_txt}**. Te guían estímulos visuales de **{calidez_txt}**.\"*\n"
final_text += f"**Parámetros calculados:** Activación: `{activacion_final:.2f}` | Positividad: `{positividad_final:.2f}` | Calidez: `{calidez_c:.2f}`\n"

from IPython.display import display, Markdown
display(Markdown(final_text))
print("-" * 80)
print("Ejecutando el recomendador en vivo...\n")

# Ejecutar recomendador físico
cmd = f'python -m src.main --bpm {BPM} --energy {Energia} --danceability {Bailabilidad} --valence {Valencia_Alegria} --color "{Color_Hex}" --presupuesto {Presupuesto_Importancia} --top 3'
if Solo_Joyas_Ocultas:
    cmd += ' --solo-joyas'
!{cmd}
````

# resumen del proyecto hasta el momento

FUSIÓN SENSORIAL + ÍNDICE DE VALOR (PySpark) - RESUMEN DEL PROYECTO
=====================================================================

1. QUÉ ESTAMOS HACIENDO
Un sistema de recomendación turística que NO pregunta al usuario qué destino quiere.
Interpreta preferencias sensoriales (música y color), las convierte en un perfil numérico
y lo compara con el perfil turístico de cada destino, calculado a partir de 208,051
reseñas reales del dataset Rest-Mex 2025. Un parámetro de "presupuesto" añade un factor
de valor/precio percibido (Índice de Valor), que permite detectar "joyas ocultas".

2. DATOS REALES (verificados sobre el archivo)
- 208,051 reseñas; columnas: Title, Review, Polarity (1-5), Town, Region, Type.
- Type: Restaurant 86,720 | Attractive 69,921 | Hotel 51,410.
- 40 destinos (Town), 19 regiones. Polarity muy sesgada: 65.6% son 5 estrellas.
- El dataset NO trae nombre de establecimiento: la unidad de recomendación es (Town, Type),
  p. ej. "Restaurantes en Tulum". Con mínimo de 150 reseñas quedan 108 unidades.
- El texto no estaba normalizado de forma uniforme (30.5% con acentos): se quitan acentos
  antes de buscar palabras.

3. DECISIÓN SOBRE SPOTIFY (investigada, no supuesta)
- Audio Features (BPM, energy, danceability, valence) y otros endpoints se cerraron para
  apps nuevas el 27-nov-2024. En febrero de 2026 hubo más restricciones (playlists:
  "tracks" pasó a "items", y el contenido puede no entregarse en playlists ajenas).
  Prueba propia: la API devolvió nombre y descripción de la playlist, pero ni "items" ni "tracks".
- La Developer Policy (mayo 2025) prohíbe usar contenido de Spotify para perfilar usuarios
  y para entrenar/ingerir en modelos de ML/IA.
- Apple Music: la API exige pertenecer al Apple Developer Program y su catálogo no expone
  tempo/energy/valence. YouTube Music: sin API oficial de características de audio.
- Conclusión: no se usa ninguna API de streaming. La música entra por:
  (a) PRESETS por género/ambiente (data/presets.json, valores típicos aproximados y editables),
  (b) parámetros manuales (BPM, energy, danceability, valence), o
  (c) CSV con esas columnas obtenido de una fuente independiente y autorizada.
- Letras de canciones: se evaluó y se DESCARTÓ para esta versión (ver sección 9).

4. ARQUITECTURA
  Entrada sensorial        Big Data (PySpark, offline)           Recomendación (Python)
  música (a/b/c) + color -> reseñas -> texto normalizado ->  ->  perfil -> intención τ ->
  (hex o imagen)            banderas por dimensión ->            afinidad coseno ->
                            agregación (Town,Type) ->            score con calidad/valor ->
                            percentiles + TF-IDF ->              top-N + explicación
                            profiles.json (108 filas)
- Parte Big Data: todo lo de build_profiles.py (Spark DataFrames, Window, joins, agregaciones,
  TF-IDF distribuido) y la fiabilidad por mitades.
- Parte recomendación: sensory.py, recommender.py, main.py, evaluate.py (Python puro sobre
  la tabla pequeña de perfiles).
- Honestidad técnica: el dataset (61 MB) cabe en memoria de una laptop. Spark se justifica
  por el procesamiento de texto a escala, por el diseño escalable (el mismo código corre
  en un clúster) y por el modelo de programación, no porque sea inmanejable.

5. PROCESAMIENTO CON PYSPARK (src/build_profiles.py)
1) Lectura del CSV con esquema explícito. 2) Normalización: minúsculas, sin acentos, solo letras.
3) Una bandera 0/1 por dimensión: ¿la reseña menciona alguna palabra del léxico?
   Dimensiones turísticas: vida nocturna, aventura, naturaleza, descanso, cultura, gastronomía,
   playa. Dimensiones de precio: valor_barato, valor_caro. (Todas con cobertura >4% medida
   en el corpus; el léxico es una hipótesis validada contra los datos.)
4) Agregación por (Town, Region, Type): n, polaridad media, tasa de mención por dimensión.
5) Calidad suavizada (bayesiana): q = (n*media + m*mu)/(n+m), m=200, mu=media global.
6) Índice de Valor: V = log((barato + a)/(caro + a)), a=0.01 (evita log(0) y división por 0).
7) Percentiles (Window): uno DENTRO de cada Type y otro GLOBAL; el perfil turístico mezcla
   ambos: t = 0.5*percentil_tipo + 0.5*percentil_global, centrado en [-1,1].
   El del tipo evita el sesgo (un restaurante siempre habla más de comida); el global
   conserva la magnitud real entre tipos.
8) "Joya oculta": valor >= 0.6, calidad >= 0.5 y popularidad (nº de reseñas) <= 0.5 en su Type.
9) TF-IDF por destino (cada Town es un documento, idf = log((N+1)/(df+1))): términos distintivos
   que se muestran como evidencia en las explicaciones.
Técnicas descartadas por no aportar valor: LDA, Word2Vec, sentimiento propio (ya existe Polarity).

6. MÓDULO SENSORIAL Y MATEMÁTICA (src/sensory.py, src/recommender.py)
Ejes en [0,1]: A = activación, P = positividad, W = calidez.
- Música: bpm_n = clip((BPM-60)/120, 0, 1); A_m = 0.35*bpm_n + 0.35*energy + 0.30*danceability;
  P_m = valence.
- Color: RGB -> HSV. Calidez W_c = (1 + cos(hue - 30°))/2 (máx. naranja, mín. azul), atenuada
  hacia 0.5 en colores poco saturados. A_c = 0.5*S + 0.3*W_c + 0.2*V; P_c = 0.5*V + 0.5*W_c.
  Imagen: paleta de 5 colores dominantes; se promedian los EJES ponderados por proporción.
- Fusión tardía: A y P = 0.5*música + 0.5*color; W solo viene del color (0.5 si no hay color).
- Intención turística: τ = M · (a, p, w) con ejes centrados a=2A-1, etc. M es una matriz
  explícita 7x3 (hipótesis heurística, ver limitaciones).
- Afinidad = (coseno(τ, t) + 1)/2.
- Calidad-valor = (1 - presupuesto)*q_norm + presupuesto*v_norm.
- Score final = 100 * afinidad^0.7 * calidad_valor^0.3.
- Explicación: dimensiones donde el usuario busca (τ>0) y el destino está sobre el promedio
  (t>0), tasas de menciones de precio, polaridad, joya oculta y términos TF-IDF.

7. EVALUACIÓN (no hay ground truth, se miden coherencia y estabilidad)
- Coherencia: un perfil rápido/cálido debe subir vida nocturna y aventura; uno lento/frío,
  descanso y naturaleza.
- Presupuesto: al subir de 0 a 1 el valor medio del top-10 debe aumentar.
- Cobertura: con 200 perfiles sensoriales aleatorios, qué fracción del catálogo llega a algún
  top-10 y cómo se reparte por Type.
- Fiabilidad por mitades (Spark): Spearman entre perfiles calculados con dos mitades aleatorias
  de las reseñas. Resultado real obtenido: vida nocturna 0.894, aventura 0.935,
  naturaleza 0.977, descanso 0.922, cultura 0.969, gastronomía 0.979, playa 0.835.
- Tabla de top-2 por preset de género (python -m src.evaluate --presets).
- Primera versión: PASS en los tres chequeos de coherencia. Hallazgo: la normalización solo
  dentro del Type hacía que los restaurantes dominaran; se corrigió con la mezcla de percentiles.
  (Completar aquí con la salida final de evaluate tras la ejecución definitiva.)

9. LIMITACIONES (declararlas en la presentación)
- La unidad es (Town, Type), no un establecimiento individual.
- La matriz M, los pesos música/color y los valores de los presets son HEURÍSTICOS
  (asociaciones cross-modales), no estimados con datos.
- El valor percibido se calcula solo con ~10% de reseñas que mencionan precio: es ruidoso
  en destinos con pocas menciones.
- El léxico es en español; hay algunas reseñas en inglés y nombres propios que se cuelan
  en los términos TF-IDF.
- La cobertura del catálogo con perfiles aleatorios no es total (algunas unidades casi
  nunca quedan en el top-10).
- Las letras se descartaron: aportan emoción (eje P) pero casi nada de ritmo/energía, requieren
  un léxico o modelo de sentimiento en español sin validar, y añaden complejidad sin
  justificación sólida. Queda como trabajo futuro.

10. MEJORAS FUTURAS
